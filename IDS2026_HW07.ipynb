The second step of CRISP-DM is data understanding. The goal is to collect, describe, explore, and verify the quality of the data to ensure it’s suitable for the analytical or modeling tasks ahead.

**a. Describe the data. Number of rows and columns, what does one row represent.**

```python
## TODO
import os
import glob
import pandas as pd
import numpy as np
# Search both local working folder and Downloads directory
file_paths = sorted(glob.glob("se_20*.csv"))
if not file_paths:
    file_paths = sorted(glob.glob(os.path.expanduser(r"~\Downloads\se_20*.csv")))
print(f"Found {len(file_paths)} yearly files:")
dfs = []
for p in file_paths:
    df_year = pd.read_csv(p, low_memory=False)
    print(f" - {os.path.basename(p)}: {df_year.shape[0]:,} rows, {df_year.shape[1]} columns")
    dfs.append(df_year)
df = pd.concat(dfs, ignore_index=True)
print("\n" + "="*50)
print(f"Total Combined Rows:    {df.shape[0]:,}")
print(f"Total Columns:          {df.shape[1]}")
print("="*50)
print("\nDataset Info:")
df.info()
display(df.head(5))
```

**Answer:**
* **Dataset Scale:** 
  The combined dataset across all six years (2021–2026) consists of **263,066 rows** and **13 columns**:
  * `se_2021.csv`: 43,004 rows
  * `se_2022.csv`: 47,639 rows
  * `se_2023.csv`: 45,972 rows
  * `se_2024.csv`: 47,097 rows
  * `se_2025.csv`: 43,653 rows
  * `se_2026.csv`: 35,701 rows (up to current reporting period)
* **What One Row Represents:**
  Every row represents a single practical driving test session taken by an applicant (`EKSAMI_SOORITAJA`), evaluated by an examiner (`EKSAMINEERIJA`) at a specific regional bureau (`BYROO`) in a given month (`KUUPAEV`), concluding with an outcome (`SEISUND`).

**c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Simple bar plot of exam outcomes (SEISUND) ilma hoiatuseta
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='SEISUND', hue='SEISUND', palette='Set2', legend=False)

plt.title('Distribution of Driving Exam Outcomes (SEISUND)')
plt.xlabel('Exam Result')
plt.ylabel('Number of Exams')

# Show percentages on the bars
total = len(df)
for p in plt.gca().patches:
    pct = f"{100 * p.get_height() / total:.1f}%"
    plt.gca().annotate(pct, (p.get_x() + p.get_width() / 2, p.get_height() / 2),
                       ha='center', color='black', fontsize=11, weight='bold')

plt.tight_layout()
plt.show()
```

**Answer:** The plot shows the overall distribution of the target variable SEISUND (SOORITATUD vs. MITTE_SOORITATUD).
The dataset is relatively balanced, with roughly half of all exams resulting in a pass and the other half in a fail.
Because the classes are fairly evenly split, a simple majority-class baseline would achieve approximately 50% accuracy. Any useful machine learning model will need to clearly outperform this 50% baseline on test data.
